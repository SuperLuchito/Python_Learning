```python
# My Solution | Runtime: 13 ms, Beats 29.56% | Memory: 12.37 MB, Beats 83.45%

s = "IX"
def romanToInt(s):
    roman_numerals = {
        "I": 1,
        "V": 5,
        "X": 10,
        "L": 50,
        "C": 100,
        "D": 500,
        "M": 1000
    }
    
    s_list = list(s)
    answer_num = 0
    
    int_num_list = []
    
    for i in range(len(s_list)):
        int_num_list.append(roman_numerals[s_list[i]])
        
    for i in range(len(int_num_list)):
        if i < len(int_num_list) - 1 and int_num_list[i] < int_num_list[i + 1]:
            answer_num -= int_num_list[i]
        else:
            answer_num += int_num_list[i]
    return answer_num

romanToInt(s)
```

```python
# Solution of AI (Gemeni 2.5 Pro) | Runtime: 11 ms, Beats 43.25% | Memory: 12.49 MB, Beats 52.73%

def romanToInt(s):
    roman_map = {
        'I': 1, 
        'V': 5, 
        'X': 10, 
        'L': 50, 
        'C': 100, 
        'D': 500, 
        'M': 1000
    }
    
    total = 0
    i = 0
    while i < len(s):
        if i + 1 < len(s) and roman_map[s[i]] < roman_map[s[i+1]]:
            total += roman_map[s[i+1]] - roman_map[s[i]]
            i += 2
        else:
            total += roman_map[s[i]]
            i += 1
    return total

result = romanToInt("MCMXCIV")
print(result)
```
