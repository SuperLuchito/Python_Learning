```python
# My Solution | Runtime: 4 ms, Beats 20.36% | Memory: 12.30 MB,Beats 99.67%

def longestCommonPrefix(strs):
    final_prefix = strs[0]
    for i in range(1, len(strs)):
        while not strs[i].startswith(final_prefix):
            final_prefix = final_prefix[:-1]
    return final_prefix

strs = ["flower", "flow", "flight"]

longestCommonPrefix(strs)
```

```python
# My Solution | Runtime: 0ms, Beats 100.00% | Memory: 12.58MB, Beats 26.33%

def longestCommonPrefix(strs):
    if not strs:
        return ""
    
    strs.sort()
    
    first = strs[0]
    last = strs[-1]
    
    i = 0
    while i < len(first) and i < len(last) and first[i] == last[i]:
        i += 1
        
    return first[:i]

result = longestCommonPrefix(["flower","flow","flight"])
print(result)
```
